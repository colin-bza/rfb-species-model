Analytical solution for first order dissolution

```python
import numpy as np
# Maximum concentration, to be calculated
c0 = 0.5

# Reaction constant
k = 0.125

# Diffusivity, to be updated
D = 1.569e-5

# Velocity / m / s
v = 1.121e-3

# The domain is 14mm long
L = 0.014

# Number of vertical elements
n_z = 100

# Vertical coordinate vector to sample solution
z = np.linspace(0, L, n_z)
```

```python
tau_ad = L/v
tau_diff = L**2/D
tau_rx = 1/k
print("tau_ad = %.2f s" % tau_ad)
print("tau_diff = %.2f s" % tau_diff)
print("tau_rx = %.2f s" % tau_rx)
```

Define analytical solution

```python
def c_an(z, c0, k, D, v):
    ''' Concentration profile in steady state for 1-D
    advection diffusion reaction equation'''

    return c0 * np.exp( - (np.sqrt(1 + (4*k*D/v**2)) - 1) * (v*z/(2*D)) )
```

```python
plt.plot(z, c_an(z, c0, k, D, v))
plt.xlabel('z')
plt.ylabel('c')
```
