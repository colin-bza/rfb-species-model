```python
import numpy as np
import pandas as pd
from pyDOE2 import lhs


n_vars = 3
n_samples = 64


lhs_normalized = lhs(n=n_vars, samples=n_samples, criterion='maximin')


def scale_column(col, min_val, max_val):
    return col * (max_val - min_val) + min_val


porosity    = scale_column(lhs_normalized[:, 0], 0.3, 0.8)
electrode_mm  = scale_column(lhs_normalized[:, 1], 2, 6)
AR            = scale_column(lhs_normalized[:, 2], 0.5, 2.1)

# Data frame
df_lhs = pd.DataFrame({
    'Porosidad': porosity,
    'Electrodo_mm': electrode_mm,
    'AR': AR
})


print(df_lhs.head(100))

#Save CSV
df_lhs.to_csv('diseño_LHS.csv', index=False)
```
